```python
import json
import time
import random
from datetime import datetime
from kafka import KafkaProducer

producer = KafkaProducer(
    bootstrap_servers="localhost:9092",
    value_serializer=lambda v: json.dumps(v).encode("utf-8")
)

orders = [
    (101, 501, "Udupi"),
    (102, 502, "Manipal"),
    (103, 503, "Mangalore"),
    (104, 501, "Udupi"),
    (105, 504, "Kundapura")
]

status_list = [
    "PLACED", "PREPARING", "OUT_FOR_DELIVERY", "DELIVERED"
]

cuisine_list = ["Indian", "Chinese", "Italian", "South Indian"]

def generate_order_event():
    order = random.choice(orders)

    return {
        "order_id": order[0],
        "restaurant_id": order[1],
        "delivery_area": order[2],
        "status": random.choice(status_list),
        "estimated_delivery_time": random.randint(10, 60),
        "cuisine": random.choice(cuisine_list),
        "event_time": datetime.now().isoformat()
    }

print("Producing Food Delivery events...")

try:
    while True:
        event = generate_order_event()
        producer.send("food-delivery-updates", event)
        producer.flush()
        print("Sent:", event)
        time.sleep(1)

except KeyboardInterrupt:
    print("Producer stopped.")

finally:
    producer.close()
```
