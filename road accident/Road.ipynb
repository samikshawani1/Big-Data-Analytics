```python
import pandas as pd

# Read datasets
df = pd.read_csv("Road_accidents.csv")
time_df = pd.read_csv("Road_accidents_time.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()
time_df.columns = time_df.columns.str.strip()

print("Road_accidents columns:")
print(df.columns.tolist())

print("\nRoad_accidents_time columns:")
print(time_df.columns.tolist())
```

```python
#1.Add column 'Total', giving sum of accidents in each month .
month_columns = [
    'JANUARY',
    'FEBRUARY',
    'MARCH',
    'APRIL',
    'MAY',
    'JUNE',
    'JULY',
    'AUGUST',
    'SEPTEMBER',
    'OCTOBER',
    'NOVEMBER',
    'DECEMBER'
]

df['Total'] = df[month_columns].sum(axis=1)

print(df.head())
```

```python
#2.Which state has highest number of accidents in year 2013? 
data_2013 = df[df['YEAR'] == 2013]

state_2013 = data_2013.groupby('STATE/UT')['Total'].sum()

print("Accidents by state in 2013:")
print(state_2013.sort_values(ascending=False))

print("\nState with highest accidents in 2013:")
print(state_2013.idxmax())

print("Number of accidents:")
print(state_2013.max())
```

```python
#3.Find the average monthly accidents for each state. 
state_average = df.groupby('STATE/UT')['Total'].mean()

print("Average monthly accidents for each state:")
print(state_average.sort_values(ascending=False))
```

```python
print(state_average.round(2).sort_values(ascending=False))
```

```python
#4.Which month has highest accidents? 
monthly_total = df[month_columns].sum()

print("Total accidents in each month:")
print(monthly_total.sort_values(ascending=False))

print("\nMonth with highest accidents:")
print(monthly_total.idxmax())

print("Number of accidents:")
print(monthly_total.max())
```

```python
#5. Which time slot (like 0-3, 3-6 etc) has more accidents? What about 'Karnataka' state? 
time_columns = [
    '0-3 hrs. (Night)',
    '3-6 hrs. (Night)',
    '6-9 hrs (Day)',
    '9-12 hrs (Day)',
    '12-15 hrs (Day)',
    '15-18 hrs (Day)',
    '18-21 hrs (Night)',
    '21-24 hrs (Night)'
]

print(time_columns)
```

```python
time_total = time_df[time_columns].sum()

print("Accidents by time slot:")
print(time_total.sort_values(ascending=False))

print("\nTime slot with highest accidents:")
print(time_total.idxmax())

print("Number of accidents:")
print(time_total.max())
```

```python
# For Karnataka
karnataka = time_df[
    time_df['STATE/UT'].str.strip().str.upper() == 'KARNATAKA'
]

karnataka_time = karnataka[time_columns].sum()

print("Karnataka accidents by time slot:")
print(karnataka_time.sort_values(ascending=False))

print("\nHighest accident time slot in Karnataka:")
print(karnataka_time.idxmax())

print("Number of accidents:")
print(karnataka_time.max())
```

```python
#6.Which state has more accidents from year 2001 to 2014? 
state_total = df.groupby('STATE/UT')['Total'].sum()

print("Total accidents from 2001 to 2014:")
print(state_total.sort_values(ascending=False))

print("\nState with highest accidents:")
print(state_total.idxmax())

print("Number of accidents:")
print(state_total.max())
```

```python
# 7.List states whose accidents number in year 2014 is less than state average from 2001 to 2014.
state_average = df.groupby('STATE/UT')['Total'].mean()

# Total accidents in 2014
accidents_2014 = (
    df[df['YEAR'] == 2014]
    .groupby('STATE/UT')['Total']
    .sum()
)

comparison = pd.DataFrame({
    'Average_2001_2014': state_average,
    'Accidents_2014': accidents_2014
})

# Select states where 2014 is less than their average
result = comparison[
    comparison['Accidents_2014'] < comparison['Average_2001_2014']
]

print("States where 2014 accidents are less than their 2001-2014 average:")
print(result)
```

```python
#8.Which month has more accidents over the year? 
month_total = df[month_columns].sum()

print("Accidents by month:")
print(month_total.sort_values(ascending=False))

print("\nMonth with highest accidents:")
print(month_total.idxmax())

print("Number of accidents:")
print(month_total.max())
```
