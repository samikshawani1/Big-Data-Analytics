```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
 
spark = SparkSession.builder \
    .appName("Students") \
    .getOrCreate()
 
df = spark.read.csv(
    "students.csv",
    header=True,
    inferSchema=True
)
 
df = df.toDF(
    "gender",
    "race",
    "parental_education",
    "lunch",
    "test_preparation",
    "math_score",
    "reading_score",
    "writing_score"
)
 
df.show(5)
```

```python
#1.Number of male and female students 
df.groupBy("gender") \
  .count() \
  .show()
```

```python
#2.List different 'race/ethnicity' 
df.select("race") \
  .distinct() \
  .show()
```

```python
#3.What are different 'parental level of education'?
df.select("parental_education") \
  .distinct() \
  .show()
```

```python
#4.How many female students scored more than 79 marks in math, whose parental level of 
#education is 'high school'? 
df.filter(
    (col("gender") == "female") &
    (col("math_score") > 79) &
    (col("parental_education") == "high school")
).count()
```

```python
#4.with records
df.filter(
    (col("gender") == "female") &
    (col("math_score") > 79) &
    (col("parental_education") == "high school")
).show()
```

```python
#5. Check whether average maths score of male or female students are high? 
df.groupBy("gender") \
  .agg(
      avg("math_score").alias("average_math_score")
  ) \
  .orderBy(desc("average_math_score")) \
  .show()
```

```python
#6. What is average reading score of male and female students? 
df.groupBy("gender") \
  .agg(
      avg("reading_score").alias("average_reading_score")
  ) \
  .show()
```

```python
#7.Whether students score depends upon 'parental level of education'? Justify your answer. 
df.groupBy("parental_education") \
  .agg(
      avg("math_score").alias("average_math"),
      avg("reading_score").alias("average_reading"),
      avg("writing_score").alias("average_writing")
  ) \
  .show()
```

```python
#8.List the records where 'test preparation course' is 'none' and scored more than 70 in maths.
df.filter(
    (col("test_preparation") == "none") &
    (col("math_score") > 70)
).show()
```
